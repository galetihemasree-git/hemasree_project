## WHAT ARE THE MOST POPULAR PRODUCT CATEGERIOUS AMONG OUR CUSTOMERS



```python
product_line = df.groupby('Attendance Rate')['Previous Grades'].sum().sort_values(ascending=False)
product_line
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=product_line.index,y=product_line.values)
plt.xlabel('Parent Education Level')
plt.ylabel('Associate')
plt.title('Master')
plt.xticks(rotation=45)
plt.show()
```

```python
payment_method=df['Passed'].value_counts()
plt.pie(payment_method.values, labels=payment_method.index, autopct='%1.1f%%')
plt.title('Results of Passed Students')
plt.show()
```

## WHAT IS THE AVERAGE ATTENDANCE RATEOF THE CLASS??

```python
sns.histplot(df['Attendance Rate'],bins=9)
plt.xlabel('Attendance Rate')
plt.ylabel('Frequency')
plt.title('Distribution of Attendance Rate')
plt.show()
```

```python
avg_attendance_rate_by_passed = df.groupby('Passed')['Attendance Rate'].mean()

plt.plot(df['Passed'], df['Attendance Rate'], marker='o', linestyle='-')
plt.xlabel('Passed')
plt.ylabel('Attendance Rate')
plt.title('Attendance Rate Over Passed')
plt.grid(True)
plt.show()
```

```python
avg_attendance_rate_by_previous_grades = df.groupby('Previous Grades')['Attendance Rate'].mean()
plt.figure(figsize=(10, 6))
plt.plot(avg_attendance_rate_by_previous_grades.index,avg_attendance_rate_by_previous_grades.values, marker='o', linestyle='-')
plt.xlabel('Previous Grades')
plt.ylabel('Average Attendance Rate')
plt.title('Average Attendance Rate by Previous Grades')
plt.grid(True)
plt.show()
```
